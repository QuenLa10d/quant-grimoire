```python
import numpy as np
import matplotlib.pyplot as plt

# Paramètres de simulation
T = 1.0       # Temps total
N = 500       # Nombre de pas
dt = T / N    # Pas de temps

# Génération des accroissements browniens (loi normale)
dW = np.random.normal(0, np.sqrt(dt), N)

# Construction de la trajectoire (somme cumulative)
W = np.insert(np.cumsum(dW), 0, 0) # W_0 = 0

# Visualisation
plt.figure(figsize=(10, 5))
plt.plot(np.linspace(0, T, N+1), W, color='darkblue', lw=1.5)
plt.title("Simulation d'un Mouvement Brownien Standard")
plt.xlabel("Temps t")
plt.ylabel("W(t)")
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
